```python
# Ejercicio 1
# 5x + 2y = 1
# -3x + 3y = 5

import numpy as np
import matplotlib.pyplot as plt
from sympy import symbols, Eq, solve

x, y = symbols('x y')

# Definir ecuaciones
ecuacion1 = Eq(5*x + 2*y, 1)
ecuacion2 = Eq(-3*x + 3*y, 5)

# Resolver sistema
sol = solve((ecuacion1, ecuacion2), (x, y))
x_sol, y_sol = sol[x], sol[y]

print("✅ Solución:")
print(f"x = {x_sol}, y = {y_sol}")

# Graficar rectas
x_vals = np.linspace(-5, 5, 400)
y1 = (1 - 5*x_vals)/2
y2 = (5 + 3*x_vals)/3

plt.figure(figsize=(6,6))
plt.plot(x_vals, y1, label="5x + 2y = 1", color="blue")
plt.plot(x_vals, y2, label="-3x + 3y = 5", color="red")
plt.scatter(float(x_sol), float(y_sol), color="black", s=80, label=f"({x_sol}, {y_sol})")

plt.axhline(0, color="gray", linewidth=0.7)
plt.axvline(0, color="gray", linewidth=0.7)
plt.legend()
plt.title("Sistema de ecuaciones")
plt.grid(True)
plt.show()
```

```python
# Ejercicio 1-b
import numpy as np
import matplotlib.pyplot as plt
from sympy import symbols, Eq, solve

x, y = symbols('x y')

# --- CAMBIA AQUÍ LAS ECUACIONES ---
# Ejercicio a) 5x+2y=1  ,  -3x+3y=5
# ecuacion1, ecuacion2 = Eq(5*x+2*y,1), Eq(-3*x+3*y,5)

# Ejercicio b) 2x+y=6  ,  4x+3y=14
ecuacion1, ecuacion2 = Eq(2*x+y,6), Eq(4*x+3*y,14)
# ---------------------------------

sol = solve((ecuacion1, ecuacion2), (x, y))
print("✅ Solución:", sol)

# Graficar
x_vals = np.linspace(-5, 5, 400)
y1 = [solve(ecuacion1, y)[0].subs(x, v) for v in x_vals]
y2 = [solve(ecuacion2, y)[0].subs(x, v) for v in x_vals]

plt.plot(x_vals, y1, "b", label=str(ecuacion1))
plt.plot(x_vals, y2, "r", label=str(ecuacion2))
plt.scatter(float(sol[x]), float(sol[y]), color="black", s=80, label=f"Sol {sol}")
plt.axhline(0, color="gray"); plt.axvline(0, color="gray")
plt.legend(); plt.grid(True); plt.show()
```

```python
# Ejercicio 11
import numpy as np
import matplotlib.pyplot as plt

# Ejercicio 11: Sistema con fracciones
print("Ejercicio 11:")
print("2(x + 4)/3 - y/2 = 9/2")
print("x + 2y - (1/3)(3x - 2) = -4/3")
print()

# Simplificar ecuaciones
# Ecuación 1: 4x - 3y = 11 → y = (4x - 11)/3
# Ecuación 2: 6y = -6 → y = -1

# Resolver con numpy
A = np.array([[4, -3],
              [0, 6]])
b = np.array([11, -6])
solucion = np.linalg.solve(A, b)

print(f"Solución: x = {solucion[0]}, y = {solucion[1]}")

# Verificación
x_val, y_val = solucion[0], solucion[1]
check1 = 2*(x_val + 4)/3 - y_val/2
check2 = x_val + 2*y_val - (1/3)*(3*x_val - 2)
print(f"Verificación: {check1:.1f} = {9/2}, {check2:.2f} = {-4/3:.2f}")

# Gráfico
x = np.linspace(-2, 6, 100)
y1 = (4*x - 11)/3  # Ecuación 1: 4x - 3y = 11
y2 = np.full_like(x, -1)  # Ecuación 2: y = -1

plt.figure(figsize=(8, 6))
plt.plot(x, y1, 'b-', label='4x - 3y = 11', linewidth=2)
plt.plot(x, y2, 'r-', label='y = -1', linewidth=2)
plt.plot(solucion[0], solucion[1], 'go', markersize=10, label=f'Solución ({solucion[0]}, {solucion[1]})')
plt.grid(True, alpha=0.3)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Sistema de Ecuaciones - Ejercicio 11')
plt.legend()
plt.xlim(-1, 5)
plt.ylim(-3, 2)
plt.axhline(y=0, color='k', linewidth=0.5)
plt.axvline(x=0, color='k', linewidth=0.5)
plt.show()
```

```python
# EJERCICIO 20
import numpy as np
import matplotlib.pyplot as plt

# Ecuación: 2x + y = 7 -> y = 7 - 2x
xx = np.linspace(-1, 5, 400)
yy = 7 - 2*xx

plt.figure(figsize=(6,6))
plt.plot(xx, yy, label="2x + y = 7")

# Pares del PDF
pares = [(0,7), (1,5), (2,3), (3,7), (4,0)]

# Graficar cada par y marcar si es solución o no
for (x, y) in pares:
    if 2*x + y == 7:
        plt.scatter(x, y, color="green", s=80, label=f"{(x,y)} ✔" if (x,y)==(0,7) else None)
    else:
        plt.scatter(x, y, color="red", s=80, marker="x", label=f"{(x,y)} ✘" if (x,y)==(3,7) else None)

plt.axhline(0, color="black", linewidth=0.5)
plt.axvline(0, color="black", linewidth=0.5)
plt.legend(); plt.grid(True)
plt.title("Ejercicio 20")
plt.show()

# Resultados en consola
print("Ejercicio 20 - Ecuación: 2x + y = 7")
for par in pares:
    x, y = par
    if 2*x + y == 7:
        print(f"{par} -> ES solución")
    else:
        print(f"{par} -> NO es solución")

print("\nConclusión: Las soluciones son (0,7), (1,5), (2,3).")
print("Los puntos de la recta son las soluciones de la ecuación.")
```

```python
# EJERCICIO 25
import numpy as np
import matplotlib.pyplot as plt

# Sistema (rectas que se cortan en (-2,1)):
#   x + y = -1   -> y = -1 - x
#   x - y = -3   -> y = x + 3
A = np.array([[1, 1],
              [1,-1]], dtype=float)
b = np.array([-1, -3], dtype=float)

sol = np.linalg.solve(A, b)
x_sol, y_sol = sol

# Mostrar resultados
print("Solución:")
print(f"x = {x_sol}, y = {y_sol}")

# Gráfica
xx = np.linspace(-8, 4, 400)
y1 = -1 - xx   # y = -1 - x
y2 = xx + 3    # y = x + 3

plt.figure(figsize=(6,6))
plt.plot(xx, y1, label="x + y = -1")
plt.plot(xx, y2, label="x - y = -3")
plt.scatter([x_sol], [y_sol])         # marca la solución
plt.annotate(f"({int(x_sol)},{int(y_sol)})", (x_sol, y_sol),
             xytext=(6,-10), textcoords='offset points')
plt.axhline(0, linewidth=0.5)
plt.axvline(0, linewidth=0.5)
plt.legend()
plt.grid(True)
plt.xlim(-8,4)
plt.ylim(-6,6)
plt.title("Ejercicio 25")
plt.show()
```
