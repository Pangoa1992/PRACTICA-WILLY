Parte 4: Mini-proyecto

1. Agrega una nueva columna llamada Crecimiento que calcule el porcentaje de
crecimiento entre Ventas_2024 y Ventas_2025.

```python
ventas = pd.DataFrame({
    "Producto": ["A", "B", "C", "D", "E"],
    "Ventas_2024": [1000, 500, 0, 200, 800],
    "Ventas_2025": [1500, 450, 50, 400, 1600]
})

def crecimiento_pct(row):
    v2024, v2025 = row["Ventas_2024"], row["Ventas_2025"]
    if v2024 == 0:
        return np.inf if v2025 > 0 else 0.0
    return (v2025 - v2024) / v2024 * 100

ventas["Crecimiento"] = ventas.apply(crecimiento_pct, axis=1)
print(ventas)
```

2. Identifica el producto con mayor crecimiento y muéstralo.

```python
print("Producto con mayor crecimiento:")
print(ventas.loc[ventas["Crecimiento"].idxmax()])
```

Parte 5: Resolución de ecuaciones

1. Ecuación de primer grado
o Implementa una función en Python que resuelva una ecuación de la forma:
ax+b=0ax + b = 0ax+b=0  
por ejemplo: 2x + 4 = 0.
o La función debe retornar el valor de x.

```python
def resolver_lineal(a, b):
    if a == 0:
        raise ValueError("a no puede ser 0 en ecuación lineal")
    return -b / a

print(resolver_lineal(2, 4))  # Ejemplo: 2x + 4 = 0
```

2. Ecuación de segundo grado
o Implementa una función en Python que resuelva una ecuación de la forma:
ax2+bx+c=0ax^2 + bx + c = 0ax2+bx+c=0  
usando la fórmula general:
x=−b±b2−4ac2ax = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}x=2a−b±b2−4ac  
o La función debe retornar las dos soluciones (reales o complejas).

```python
import cmath

def resolver_cuadratica(a, b, c):
    if a == 0:
        if b == 0:
            raise ValueError("No es válida")
        return (-c / b,)
    disc = b**2 - 4*a*c
    sqrt_disc = cmath.sqrt(disc)
    x1 = (-b + sqrt_disc) / (2*a)
    x2 = (-b - sqrt_disc) / (2*a)
    return (x1, x2)

print(resolver_cuadratica(1, -5, 6))   # Ejemplo: x² - 5x + 6 = 0
print(resolver_cuadratica(1, 0, 1))    # Ejemplo: x² + 1 = 0
```
