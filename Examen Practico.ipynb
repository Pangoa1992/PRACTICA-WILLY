```python
import pandas as pd

data = {
    "Nombre": ["Ana", "Luis", "María", "Pedro", "Sofía"],
    "Edad": [20, 21, 19, 22, 20],
    "Nota_Matemáticas": [18, 15, 16, 12, 19],
    "Nota_IA": [17, 14, 18, 13, 20]
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df.select_dtypes(include=[np.number]).mean())
```

```python
print(df[df["Nota_IA"] > 15])
```

```python
df["Promedio"] = df[["Nota_Matemáticas", "Nota_IA"]].mean(axis=1)
print(df)
```

```python
print(df.sort_values(by="Promedio", ascending=False))
```

```python
def stats_list(nums):
    arr = np.array(nums, dtype=float)
    return arr.max(), arr.min(), arr.mean()

ejemplo = [3, 7, 1, 9, 4]
print("Mayor, menor y promedio:", stats_list(ejemplo))
```

```python
def sigmoid(x):
    x = np.array(x, dtype=float)
    return 1 / (1 + np.exp(-x))

x = np.array([-2, -1, 0, 1, 2])
print("Sigmoide aplicada:", sigmoid(x))
```

```python
def clasificar_estado(prom):
    return "Aprobado" if prom >= 14 else "Desaprobado"

df["Estado"] = df["Promedio"].apply(clasificar_estado)
print(df)
```

```python
ventas = pd.DataFrame({
    "Producto": ["A", "B", "C", "D", "E"],
    "Ventas_2024": [1000, 500, 0, 200, 800],
    "Ventas_2025": [1500, 450, 50, 400, 1600]
})

def crecimiento_pct(row):
    v2024, v2025 = row["Ventas_2024"], row["Ventas_2025"]
    if v2024 == 0:
        return np.inf if v2025 > 0 else 0.0
    return (v2025 - v2024) / v2024 * 100

ventas["Crecimiento"] = ventas.apply(crecimiento_pct, axis=1)
print(ventas)
```

```python
print("Producto con mayor crecimiento:")
print(ventas.loc[ventas["Crecimiento"].idxmax()])
```

```python
def resolver_lineal(a, b):
    if a == 0:
        raise ValueError("a no puede ser 0 en ecuación lineal")
    return -b / a

print(resolver_lineal(2, 4))  # Ejemplo: 2x + 4 = 0
```

```python
import cmath

def resolver_cuadratica(a, b, c):
    if a == 0:
        if b == 0:
            raise ValueError("No es válida")
        return (-c / b,)
    disc = b**2 - 4*a*c
    sqrt_disc = cmath.sqrt(disc)
    x1 = (-b + sqrt_disc) / (2*a)
    x2 = (-b - sqrt_disc) / (2*a)
    return (x1, x2)

print(resolver_cuadratica(1, -5, 6))   # Ejemplo: x² - 5x + 6 = 0
print(resolver_cuadratica(1, 0, 1))    # Ejemplo: x² + 1 = 0
```
